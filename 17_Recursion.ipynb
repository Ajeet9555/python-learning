```python
def multiply(a, b):
    result = 0

    for i in range(b):
        result = result + a

    print(result)

multiply(3,4)
```

```python
# By recursion
def mul(a,b):
    if b == 1:
        return a
    else:
        return a + mul(a, b-1)

print(mul(3,4))
```

```python
def fact(num):
    if num == 1:
        return 1
    else:
        return num * fact(num-1)

print(fact(5))
```

```python
# Palindrome by recursion
def palin(text):
    if len(text) <= 1:
        print('Palindrome')
    else:
        if text[0] == text[-1]:
            palin(text[1:-1])
        else:
            print('Not a palindrome')

palin('madam')
palin('malayalam')
palin('python')
palin('abba')
```

```python
# Rabbit Problem(Fibonacci)
def fib(m):
    if m == 0 or m == 1:
        return 1
    else:
        return fib(m-1) + fib(m-2)

print(fib(36))
```

```python
import time
def fib(m):
    if m == 0 or m == 1:
        return 1
    else:
        return fib(m-1) + fib(m-2)

start = time.time()
print(fib(36))
print(time.time()-start)
```

```python
# Dynamic Programming 
import time
def memo(m,d):
    if m in d:
        return d[m]
    else:
        d[m] = memo(m-1,d) + memo(m-2,d)
        return d[m]

start = time.time()
d = {0:1,1:1}
print(memo(50,d))
print(time.time()-start)
```
