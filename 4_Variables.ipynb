```python
#Dynamic Binding ----> Python is a dynamically typed language. It means that you can reassign a variable to a different data type.
#Static Binding ----> In statically typed languages, once a variable is assigned a data type, it cannot be changed. You would need to declare a new variable if you want to store a different type of data.
name = "Ajeet Singh"
print(name)
name = "Ajeet"
print(name)
```

```python
# Special syntax
a = 5; b = 6; c = 7
print(a, b, c)
print(a)
print(b)
print(c)
```

```python
a, b, c = 5, 6, 7
print(a)
print(b)
print(c)
```
