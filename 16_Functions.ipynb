```python
def palindrome(number):
    number = str(number)

    if number == number[::-1]:
        return "Palindrome"
    else:
        return "Not Palindrome"


num = int(input("Enter a number: "))
result = palindrome(num)
print(result)
```

```python
num = int(input("Enter a number: "))

original = num
digits = len(str(num))

total = 0

while num > 0:
    digit = num % 10
    total = total + digit ** digits
    num = num // 10

if total == original:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

```python
def armstrong(num):
    original = num
    digits = len(str(num))
    total = 0

    while num > 0:
        digit = num % 10
        total += digit ** digits
        num //= 10

    return total == original


num = int(input("Enter a number: "))

if armstrong(num):
    print("Armstrong Number")
else:
    print("Not Armstrong Number")
```
