### Ascending substrings from a string

```python
# Concept of Indexing
c = "Hello"
print(c)
```

```python
print(c[0])
```

```python
# Types of Indexing 
# Positive Indexing
# Negative Indexing
print(c[-4])
print(c[1])
```

```python
# Slicing
c = "Hello world"
print(c[0:5])
```

```python
print(c[2:])
```

```python
print(c[2:6:2])
```

```python
print(c[:])
```

```python
print(c[0:8:3])
```

```python
print(c[0:6:-1])
```

```python
print(c[-5:-1:2])
```

```python
print(c[::-1]) # reverse of string 
```
