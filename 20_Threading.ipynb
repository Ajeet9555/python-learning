```python
from time import sleep,time
import threading


start_time = time()
def something(id):
    print("Going to sleep.....")
    sleep(1)
    print(f"Woken up.....{id}")

t1 = threading.Thread(target=something, args=[0])
t2 = threading.Thread(target=something, args=[1])

t1.start()
t2.start()

t1.join()
t2.join()
end_time = time()

print(f"Main thread Ended in {end_time-start_time} seconds")
```
