# operators

```python
# Arithmatic Operators 
x = 5
y = 2
print(x + y)
print(x - y)
print(x * y)
print(x / y)
print(x // y)
print(x % y)
print(x ** y)
```

```python
# Comprison Operators
x = 5
y = 2
print(x > y)
print(x < y)
print(x >= y)
print(x <= y)
print(x == y)
print(x != y)
```

```python
s = "Hello"
print(s * 3, end=" ")
```

```python
s = "Hello "
print(s * 3, end="")
```

```python
#Alternative Method
s = "Hello"
print((s + " ") * 3)
```

```python
s = "Hello"
print(" ".join([s] * 3))
```

```python
#Logical Operators
x = True
y = False
print(x and y)
print(x or y)
print(not x)
print(not y)
```

```python
#Bitwise Operators
x = 5
y = 3
print(x & y) #x = 0101, y = 0011, x & y = 0001
print(x | y) #x = 0101, y = 0011, x | y = 0111
print(x ^ y) #x = 0101, y = 0011, x ^ y = 0110
print(~x) #x = 0101, ~x = 1010
print(x << 2) #x = 0101, x << 2 = 010100
print(x >> 2) #x = 0101, x >> 2 = 0001
```

```python
#Assignment Operators
a = 3
print(a)
a += 2
print(a)
a -= 2
print(a)
a *= 2
print(a)
```

```python
#Identity Operators
a = 3 
b = 3
print(a is b)
```

```python
a = [1, 2, 3]
b = [1, 2, 3]
print(a is b)
```

```python
a = "Hello-world"
b = "Hello-world"

print(a is b)
```

```python
#Membership Operators
s = "Hello World"
print("Hello" in s)
```
