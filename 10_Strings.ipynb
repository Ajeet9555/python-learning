```python
# Slicing 
c = "Hello World"
print(c[0:5])  # Characters from index 0 to 4
print(c[6:11])  # Characters from index 6 to 10
```

```python
print(c[2:])  # Characters from index 2 to end
print(c[:5])  # Characters from start to index 4
print(c[-5:-2])  # Characters from index -5 to -3
print(c[:4])  # Characters from start to index 3
```

```python
print(c[2:6:2])  # Characters from index 2 to 5 with step 2
```

```python
print(c[0:8:3])  # Characters from index 0 to 7 with step 3
```

```python
print(c[0:6:-1])  # Characters from index 0 to 5 in reverse order
```

```python
print(c[::-1])  # Characters from end to start in reverse order
```
