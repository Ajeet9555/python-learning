```python
b = lambda x: 'Even' if x%2 == 0 else 'Odd'
b(3)
```

```python
def return_sum(func,L):
    result = 0
    for i in L:
        if func(i):
            result = result +i

    return result

L = [11,14,21,23,56,78,45,29,28]
x = lambda x: x%2 == 0
y = lambda x: x%2 != 0
z = lambda x: x%3 == 0

print(return_sum(x,L))
print(return_sum(y,L))
print(return_sum(z,L))
```

```python
# Map / Filter / Reduce
l = [1, 2, 3, 4, 5, 6, 7]

# map(lambda x: x*2, l)
list(map(lambda x: x*2, l))
```

```python
list(filter(lambda x: x>4,l))
```

```python
import functools
functools.reduce(lambda x,y: x + y,l)
```

```python
d = [12, 34, 56, 11, 21, 58]
functools.reduce(lambda x, y: x if x > y else y,d)
```

```python
# List Comprehension
d = [12, 34, 56, 11, 21, 58]
a1 = [item * 2 for item in d]
print(a1)
```
