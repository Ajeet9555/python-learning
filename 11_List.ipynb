```python
# List in python are mutable 
L1 = [1, 2, 3, 4, 5]
L1[-1] = 10 # Changing last element to 10
print(L1) # Output: [1, 2, 3, 4, 10]
```

```python
# Add / Append / Extend / Insert
L1.append(6) # Adding 6 at the end of the list
print(L1) # Output: [1, 2, 3, 4, 10, 6]
L1.extend([7, 8, 9]) # Adding multiple elements at the end of the list
print(L1) # Output: [1, 2, 3, 4, 10, 6, 7, 8, 9]
L1.insert(0, 0) # Adding 0 at the beginning of the list
print(L1) # Output: [0, 1, 2, 3, 4, 10, 6, 7, 8, 9]
```

```python
# pop / remove / del
L1.pop() # Removes last element
print(L1)
L1.remove(10) # Removes first occurrence of 10
print(L1)
del L1[0] # Removes first element
print(L1)
```
