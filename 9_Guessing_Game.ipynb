```python
import random 
jackpot = random.randint(1,100)
counter = 1
guess = int(input("Chal guess kar"))
while guess != jackpot:
    if guess < jackpot:
        print("guess higher")
    else:
        print("guess Lower")

    guess = int(input("Chal guess kar"))
    counter += 1

print("Sahi Jawab")
print("You took", counter, "attempts")
```

```python
n = 5
for i in range(n):
    for j in range(2*n-1):
        if (i == 0 or i == n-1 or j == 0 or j == 2*n-2 or i+j == n-1 or i-j == n-1 + n):
            print("*", end = " ")
        else:
            print(" ", end = " ")
    print()
```
