```python
#nested loop
rows = int(input("Enter the number of rows"))

for i in range(1,rows +1):
    for j in range(0,i):
        print("*", end = " ")
    print("")
```

```python
#Break Statement
for i in range(1,11):
    if i == 5:
        break
    print(i)
```

```python
#Continue 
for i in range(1,11):
    if i == 5:
        continue
    print(i)
    print("Ajeet Rajput")
```
