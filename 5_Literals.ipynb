## Literals
### Literals is a raw data given in a variables. In Python, there are various types of literals they are as follows:

- Numeric Literals 
- String Literals 
- Boolean Literals 
- Speacial Literals 

```python
a = 0b1010 #Binary literal for 10 in decimal
b = 100 #Decimal literal for 100
c = 0o310 #Octal literal for 200 in decimal
d = 0x12c #Hexadecimal literal for 300 in decimal

#float literals
float_1 = 10.5
float_2 = 1.5e2
float_3 = 1.5e-3

#Complex literals
x = 3.14j

print(a, b, c, d)
print(float_1, float_2, float_3)
print(x, x.imag, x.real)
```

### 2. String literals

```python
string = "this is a Python"
strings = "this is a Python"
char = 'P'
multiline_str = """This is a multiline string
that spans multiple lines."""
unicode = u"\U0001F600" # Unicode literal for 😀 emoji
raw_str = r"raw \n string"

print(string)
print(strings)
print(char)
print(multiline_str)
print(unicode)
print(raw_str)
```

```python
# Boolean literals
a = True + 4
b = False + 10

print("a:", a)  # Output: a: 5
print("b:", b)  # Output: b: 10
```

```python
# Special literals
a = None
print(a)  # Output: None
```

```python
a = 34 + 56
print(a)  # Output: 90
```
