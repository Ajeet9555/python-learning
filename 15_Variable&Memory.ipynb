```python
# Mutability: Mutability refers to the ability of change into the address of variable.
```

```python
import matplotlib.pyplot as plt

# Data
categories = ["Mutable", "Immutable"]
values = [4, 8]

# Create two boxes
plt.figure(figsize=(8, 4))

plt.bar(categories, values)

# Add data types inside/above the boxes
plt.text(
    0, 2,
    "list\n dict\n set\n bytearray",
    ha="center",
    va="center",
    fontsize=12
)

plt.text(
    1, 4,
    "int\n float\n complex\n bool\n str\n tuple\n frozenset\n bytes",
    ha="center",
    va="center",
    fontsize=12
)

plt.ylabel("Number of Data Types")
plt.title("Python Data Types: Mutability")

# Hide y-axis numbers because we only want two boxes
plt.yticks([])

plt.show()
```
